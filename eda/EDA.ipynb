```python
import pandas as pd

# Load the training dataset
train_df = pd.read_csv('train.csv')

# Show dataset shape and first few rows
print("Training dataset shape:", train_df.shape)
train_df.head()
```

```python
# Check distribution of the target variable
target_distribution = train_df['target'].value_counts()
print("Target distribution:")
print(target_distribution)

# Optional: Plot distribution
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
target_distribution.plot(kind='bar')
plt.xticks([0, 1], ['Non-Disaster', 'Disaster'])
plt.ylabel('Number of Tweets')
plt.title('Target Distribution')
plt.show()
```

```python
# Add text length column
train_df['text_length'] = train_df['text'].apply(len)

# Plot length distribution by class
plt.figure(figsize=(10, 6))
train_df[train_df['target'] == 0]['text_length'].plot(kind='hist', bins=50, label='Non-Disaster', alpha=0.6)
train_df[train_df['target'] == 1]['text_length'].plot(kind='hist', bins=50, label='Disaster', alpha=0.6)
plt.xlabel('Tweet Text Length')
plt.ylabel('Number of Tweets')
plt.legend()
plt.title('Tweet Length Distribution by Class')
plt.show()
```

```python
from wordcloud import WordCloud

# Combine all text into one string
all_text = " ".join(text for text in train_df['text'])

# Generate word cloud
wordcloud = WordCloud(width=800, height=400, background_color='white').generate(all_text)

# Show the word cloud
plt.figure(figsize=(10, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Overall Word Cloud')
plt.show()
```
